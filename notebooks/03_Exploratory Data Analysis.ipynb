```python
# Cargar los datos

customers = pd.read_csv("../data/sql/customers_sql.csv")
products = pd.read_csv("../data/sql/products_sql.csv")
orders = pd.read_csv("../data/sql/orders_sql.csv")
```

## Análisis univariante

El análisis univariante permite estudiar el comportamiento de cada variable de forma individual.

En esta sección se analizará la distribución de las principales variables numéricas para comprender mejor las características del conjunto de datos.

### Distribución de la edad de los clientes

La edad es una de las principales variables demográficas del proyecto.

Analizar su distribución permite identificar el perfil general de los clientes y detectar posibles concentraciones en determinados rangos de edad.

```python
plt.figure()

sns.histplot(
    data=customers,
    x="age",
    bins=15,
    color="#4C72B0",
    edgecolor="white",
    alpha=0.9
)

plt.title("Distribución de la edad de los clientes")
plt.xlabel("Edad")
plt.ylabel("Número de clientes")

plt.tight_layout()
plt.show()
```

### Observaciones

La distribución de la edad de los clientes presenta un comportamiento relativamente uniforme entre los **18 y los 80 años**.

No se observan concentraciones especialmente marcadas en ningún intervalo de edad, lo que indica que la base de clientes está distribuida de forma equilibrada a lo largo de todo el rango analizado.

Asimismo, no se aprecian valores atípicos ni irregularidades que sugieran problemas en la calidad de esta variable.

### Distribución de las valoraciones

La variable **rating** representa la valoración otorgada por los clientes a cada pedido.

Analizar su distribución permite conocer cómo se reparten las puntuaciones registradas y detectar posibles concentraciones en determinados niveles de satisfacción.

```python
plt.figure()

sns.countplot(
    data=orders,
    x="rating",
    color="#4C72B0"
)

plt.title("Distribución de las valoraciones")
plt.xlabel("Valoración")
plt.ylabel("Número de pedidos")

plt.tight_layout()
plt.show()
```

### Observaciones

La distribución de las valoraciones muestra que la puntuación **3** es la más frecuente, concentrando una parte importante de los pedidos registrados.

Las valoraciones **2** y **5** presentan una frecuencia similar, mientras que las puntuaciones **1** y **4** son las menos habituales.

En conjunto, la distribución refleja un predominio de valoraciones intermedias, sin una concentración clara en las puntuaciones más altas o más bajas.

### Distribución de la cantidad de productos por pedido

La variable **quantity** representa el número de unidades adquiridas en cada pedido.

Analizar su distribución permite conocer si los clientes tienden a realizar compras de pocas unidades o si, por el contrario, predominan los pedidos de mayor volumen.

```python
plt.figure()

sns.countplot(
    data=orders,
    x="quantity",
    color="#4C72B0"
)

plt.title("Distribución de la cantidad de productos por pedido")
plt.xlabel("Cantidad")
plt.ylabel("Número de pedidos")

plt.tight_layout()
plt.show()
```

### Observaciones

La distribución de la cantidad de productos por pedido es relativamente uniforme entre **1 y 10 unidades**.

No se observan diferencias significativas entre las distintas cantidades, lo que indica que los pedidos se distribuyen de forma equilibrada a lo largo de todo el rango disponible.

Este comportamiento sugiere que el conjunto de datos no presenta una concentración excesiva en pedidos de pocas o muchas unidades, proporcionando una muestra representativa para el análisis posterior.

### Distribución del precio unitario

La variable **unit_price** representa el precio de cada unidad adquirida en un pedido.

Analizar su distribución permite comprender cómo se reparten los precios de los productos y detectar posibles concentraciones o valores extremos.

```python
plt.figure()

sns.histplot(
    data=orders,
    x="unit_price",
    bins=20,
    color="#4C72B0",
    edgecolor="white",
    alpha=0.9
)

plt.title("Distribución del precio unitario")
plt.xlabel("Precio unitario (€)")
plt.ylabel("Número de pedidos")

plt.tight_layout()
plt.show()
```

### Observaciones

La distribución del precio unitario muestra una concentración de pedidos en determinados niveles de precio, en lugar de una distribución continua.

Este comportamiento es coherente con la estructura del catálogo de productos, ya que un número reducido de productos comparte precios unitarios específicos.

No se observan valores anómalos fuera del rango esperado, aunque sí una clara agrupación de pedidos alrededor de los distintos niveles de precio disponibles.

### Distribución del importe total de los pedidos

La variable **total_price** representa el importe total de cada pedido realizado por los clientes.

Analizar su distribución permite comprender cómo se reparten los importes de compra e identificar posibles concentraciones o valores extremos.

```python
plt.figure()

sns.histplot(
    data=orders,
    x="total_price",
    bins=20,
    color="#4C72B0",
    edgecolor="white",
    alpha=0.9
)

plt.title("Distribución del importe total de los pedidos")
plt.xlabel("Importe total (€)")
plt.ylabel("Número de pedidos")

plt.tight_layout()
plt.show()
```

### Observaciones

La distribución del importe total de los pedidos presenta una **asimetría positiva**, con una mayor concentración de pedidos en los importes más bajos y una disminución progresiva de su frecuencia a medida que aumenta el valor de la compra.

También se observan algunos pedidos de importe elevado que, aunque poco frecuentes, amplían el rango de la distribución.

Este comportamiento sugiere la existencia de una mayoría de compras de importe moderado junto con un reducido número de pedidos de mayor valor, aspecto que se analizará con más detalle en la sección de detección de valores atípicos.

### Distribución del importe de los complementos

La variable **addon_total** representa el importe correspondiente a los productos o servicios adicionales adquiridos junto con cada pedido.

Analizar su distribución permite conocer cómo se reparten estos importes y detectar posibles concentraciones o valores extremos.

```python
plt.figure()

sns.histplot(
    data=orders,
    x="addon_total",
    bins=20,
    color="#4C72B0",
    edgecolor="white",
    alpha=0.9
)

plt.title("Distribución del importe de los complementos")
plt.xlabel("Importe de los complementos (€)")
plt.ylabel("Número de pedidos")

plt.tight_layout()
plt.show()
```

### Observaciones

La distribución del importe de los complementos presenta una marcada **asimetría positiva**, con una elevada concentración de pedidos en los importes más bajos.

A medida que aumenta el importe de los complementos, la frecuencia de los pedidos disminuye progresivamente, observándose únicamente un reducido número de compras con importes elevados.

Este comportamiento sugiere que los productos o servicios adicionales suelen representar una parte reducida del importe total de los pedidos, aunque existen algunos casos con un gasto considerable que serán revisados en la sección de detección de valores atípicos.

## Detección de valores atípicos

Los valores atípicos (outliers) son observaciones que se alejan significativamente del comportamiento general de los datos.

En esta sección se utilizarán diagramas de caja (*boxplots*) para identificar la posible presencia de valores extremos en las principales variables numéricas del proyecto.

### Edad de los clientes

Se analiza la distribución de la edad mediante un diagrama de caja para comprobar la existencia de posibles valores atípicos y evaluar la dispersión de esta variable.

```python
plt.figure()

sns.boxplot(
    data=customers,
    x="age",
    color="#4C72B0"
)

plt.title("Boxplot de la edad de los clientes")
plt.xlabel("Edad")

plt.tight_layout()
plt.show()
```

### Observaciones

El diagrama de caja muestra que la variable **age** no presenta valores atípicos.

La mediana se sitúa aproximadamente en los **49 años**, mientras que el rango intercuartílico refleja una distribución equilibrada de las edades de los clientes.

En conjunto, la variable presenta un comportamiento estable y no requiere ningún tratamiento adicional antes de su utilización en los análisis posteriores.

### Distribución de los importes

A continuación se analizan mediante diagramas de caja las variables **total_price** y **addon_total**.

El objetivo es evaluar la dispersión de los importes e identificar la posible presencia de valores atípicos que puedan influir en el análisis posterior.

```python
# Boxplot del importe total

plt.figure()

sns.boxplot(
    data=orders,
    x="total_price",
    color="#4C72B0"
)

plt.title("Boxplot del importe total de los pedidos")
plt.xlabel("Importe total (€)")

plt.tight_layout()
plt.show()
```

```python
# Boxplot del importe de los complementos

plt.figure()

sns.boxplot(
    data=orders,
    x="addon_total",
    color="#4C72B0"
)

plt.title("Boxplot del importe de los complementos")
plt.xlabel("Importe de los complementos (€)")

plt.tight_layout()
plt.show()
```

### Observaciones

Los diagramas de caja muestran que ambas variables presentan una distribución asimétrica hacia la derecha, aunque con comportamientos diferentes.

- **total_price** presenta una dispersión considerable y únicamente se observan unos pocos valores atípicos de importe muy elevado. Estos casos representan pedidos excepcionales, pero no parecen afectar al comportamiento general de la distribución.

- **addon_total** muestra un mayor número de valores atípicos. La mayoría de los pedidos presentan importes reducidos en complementos, mientras que un grupo más pequeño registra gastos significativamente superiores.

En conjunto, los valores atípicos identificados son coherentes con la naturaleza del negocio y no sugieren errores en los datos, por lo que se mantendrán para los análisis posteriores.

## Relaciones entre variables

Además de estudiar cada variable de forma individual, resulta útil analizar la relación existente entre las principales variables numéricas del conjunto de datos.

Para ello, se calculará la matriz de correlación y se representará mediante un mapa de calor (*heatmap*), facilitando la identificación de posibles relaciones lineales entre las variables.

```python
# Variables numéricas para el análisis de correlación

numeric_columns = [
    "rating",
    "unit_price",
    "quantity",
    "total_price",
    "addon_total"
]

correlation = orders[numeric_columns].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Matriz de correlación de las variables numéricas")

plt.tight_layout()
plt.show()
```

### Observaciones

La matriz de correlación muestra que las relaciones más relevantes se producen entre **unit_price**, **quantity** y **total_price**.

En concreto:

- Existe una correlación positiva moderada-alta entre **unit_price** y **total_price** (0,68).
- También se observa una correlación positiva moderada-alta entre **quantity** y **total_price** (0,65).

Estas relaciones son coherentes con la naturaleza del negocio, ya que el importe total de un pedido depende directamente del precio unitario y de la cantidad de productos adquiridos.

El resto de variables presentan correlaciones débiles o prácticamente nulas, lo que indica una baja dependencia lineal entre ellas.
